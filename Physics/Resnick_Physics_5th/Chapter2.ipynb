$$\mathbf{a} = 5.2\hat{i}$$

## 9. Magnitude and angle

$$\mathbf{a} = 4.0\hat{i} - 3.0\hat{j}$$
$$\mathbf{b} = 6.0\hat{i} + 8.0\hat{j}$$

```python
def vector_info(v):
    magnitude = np.linalg.norm(v)
    angle = np.rad2deg(np.arctan2(v[1], v[0]))

    return magnitude, angle


a = np.array([4.0, -3.0])
b = np.array([6.0, 8.0])

a_mag, a_angle = vector_info(a)
b_mag, b_angle = vector_info(b)
c_mag, c_angle = vector_info(a + b)
d_mag, d_angle = vector_info(b - a)
e_mag, e_angle = vector_info(a - b)

print(f"Vector = {a[0]:.1f}î + {a[1]:.1f}ĵ ; Magnitude : {a_mag:.2f} ; Angle : {round(a_angle, 2)}° ")
print(f"Vector = {b[0]:.1f}î + {b[1]:.1f}ĵ ; Magnitude : {b_mag:.2f} ; Angle : {round(b_angle, 2)}° ")
print(f"Vector = {c[0]:.1f}î + {c[1]:.1f}ĵ ; Magnitude : {c_mag:.2f} ; Angle : {round(c_angle, 2)}° ")
print(f"Vector = {d[0]:.1f}î + {d[1]:.1f}ĵ ; Magnitude : {d_mag:.2f} ; Angle : {round(d_angle, 2)}° ")
print(f"Vector = {e[0]:.1f}î + {e[1]:.1f}ĵ ; Magnitude : {e_mag:.2f} ; Angle : {round(e_angle, 2)}° ")
```

## 15. Missle detection

## 17. Position, velocity and acceleration

$$
\vec{r}
=
\left[(2\,\mathrm{m/s^3})t^3-(5\,\mathrm{m/s})t\right]\hat{i}
+
\left[(6\,\mathrm{m})-(7\,\mathrm{m/s^4})t^4\right]\hat{j}
$$
$$ t = 2\,\mathrm{s}

```python
import sympy as sp

t = sp.symbols('t')

r = sp.Matrix([
    2*t**3 - 5*t,
    6 - 7*t**4
])

v = r.diff(t)
a = v.diff(t)

r_at_2 = r.subs(t, 2)
v_at_2 = v.subs(t, 2)
a_at_2 = a.subs(t, 2)

print(f"r(2) = {r_at_2[0]} î + {r_at_2[1]} ĵ")
print(f"Velocity = {v_at_2[0]} î {v_at_2[1]} ĵ")
print(f"Acceleration = {a_at_2[0]} î {a_at_2[1]} ĵ")
```
