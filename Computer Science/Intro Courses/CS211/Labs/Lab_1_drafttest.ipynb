Point class 

```python
public class Point {
    private int x;
    private int y; 
    
    public Point(int x, int y) {
        this.x = x;
        this.y = y;
    }

    public int getX() {
        return this.x;
    }
    public int getY() {
        return this.y;
    }
    //Code goes here

    // a)
    public int quadrant() {
        if (x > 0 && y > 0) {
            return 1;
        } else if (x < 0 && y > 0) {
            return 2;
        } else if (x < 0 && y < 0) {
            return 3;
        } else if (x > 0 && y < 0) {
            return 4;
        } else {
            return 0;
        }
    }
    
    // b)
    public void flip() {
        int tem = this.x;
        this.x = this.y * (-1);
        this.y = tem * (-1); 
    }

    // c)
    public int manhattanDistance(Point point) {
        int dx = this.x - point.getX();
        int dy = this.y - point.getY();
        return Math.abs(dx) + Math.abs(dy);
    }

    // d)
    public boolean isVertical(Point point) {
        return (this.x == point.getX());
    }

    // e)
    public double slope(Point point) {
        double dx = point.getX() - this.x;
        double dy = point.getY() - this.y;
        if (dx == 0) {
            throw new IllegalArgumentException("Two points are vertical or at the same location");
        }
        return (dy / dx);
    }

    // f)
    public boolean isCollinear(Point point1, Point point2) {
        double slope1 = (double) (point1.getY() - this.y) / (point1.getX() - this.x);
        double slope2 = (double) (point2.getY() - this.y) / (point2.getX() - this.x);
        double slope3 = (double) (point2.getY() - point1.getY()) / (point2.getX() - point1.getX());

        return Double.compare(slope1, slope2) == 0 && Double.compare(slope2, slope3) == 0;
    }

}
```

#### interface PaymentMethod

```python
public interface PaymentMethod {
    boolean processPayment(double amount);
    void printReceipt();
}
```

#### Classes

```python
public class CreditCardPayment implements PaymentMethod {
    private String cardNumber;
    private String cardHolderName;

    public CreditCardPayment(String cardNumber, String cardHolderName) {
        this.cardNumber = cardNumber;
        this.cardHolderName = cardHolderName;
    }

    public boolean processPayment(double amount) {
        System.out.println("Processing credit card payment...");
        return true;
    }

    public void printReceipt() {
        String lastfour = cardNumber.substring(cardNumber.length() - 4);
        String stars = "**** "; 
        
        System.out.println("Receipt:  " + stars.repeat(3) + lastfour);
        System.out.println("Cardholder: " + cardHolderName);
        System.out.println();

    }
}
```

```python
public class DebitCardPayment implements PaymentMethod {
    private String cardNumber;
    private String bankName;

    public DebitCardPayment(String cardNumber, String bankName) {
        this.cardNumber = cardNumber;
        this.bankName = bankName;
    }

    public boolean processPayment(double amount) {
        System.out.println("Processing credit card payment...");
        return true;
    }

    public void printReceipt() {
        String lastfour = cardNumber.substring(cardNumber.length() - 4);
        String stars = "**** "; 
        
        System.out.println("Receipt:  " + stars.repeat(3) + lastfour);
        System.out.println("Bank: : " + bankName);
        System.out.println();
    }
}
```

```python
public class PayPalPayment implements PaymentMethod {
    private String email;

    public PayPalPayment(String email) {
        this.email = email;
    }

    public boolean processPayment(double amount) {
        System.out.println("Processing PayPal payment for " + email + "...");
        return true;
    }

    public void printReceipt() {
        System.out.println("Receipt: PayPal Account: " + email);
    }
}
```

#### Main class

```python
public class Main {
    public static void main(String[] args) {
        PaymentMethod[] payments = { 
            new CreditCardPayment("1234123412341234", "John Doe"),
            new DebitCardPayment("5678567856785678", "XYZ Bank"),
            new PayPalPayment("john@example.com")
        };

        int i = 1;
        for (PaymentMethod payment : payments) {
            payment.processPayment(i * 50);
            payment.printReceipt();
            i += i;
        }
    }   
}
```
