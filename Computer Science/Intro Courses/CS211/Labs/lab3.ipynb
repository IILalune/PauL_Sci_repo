### 4. splitStack

```python
public static void splitStack(Stack<Integer> stack) {
    Queue<Integer> queue = new LinkedList<>();

    // Transfer from stack to aux queue
    while (!stack.isEmpty()) {
        queue.add(stack.pop());
    }

    // Go through queue first, get all the negative
    int size = queue.size();
    for (int i = 0; i < size; i++) {
        int num = queue.remove();
        if (num < 0) {
            stack.push(num);
        } else {
            queue.add(num);
        }
    }

    // Add the rest of the queue to the stack
    while (!queue.isEmpty()) {
        stack.add(queue.remove());
    }
}
```

```python
Stack<Integer> stack = new Stack<>();
stack.push(5);
stack.push(-10);
stack.push(15);
stack.push(0);
stack.push(-5);
stack.push(100);
System.out.println("Original Stack: " + stack);
splitStack(stack);
System.out.println("Split Stack: " + stack);
```

### 5. copyStack

```python
public static Stack<Integer> copyStack(Stack<Integer> stack) {
    Queue<Integer> queue = new LinkedList<>();
    Stack<Integer> copy = new Stack<>();
    
    // Transfer fro original to copy stack
    while (!stack.isEmpty()) {
        copy.push(stack.pop());
    }
    // Transfer from copy stack to queue 
    while (!copy.isEmpty()) {
        queue.add(copy.pop());
    }
    //Restore the original stack and make a copy
    while (!queue.isEmpty()) {
        int num = queue.remove();
        stack.push(num);
        copy.push(num);
    }    
    //Return a copy
    return copy;
}
```

```python
Stack<Integer> stack = new Stack<>();
stack.push(5);
stack.push(-10);
stack.push(15);
stack.push(0);
stack.push(-5);
stack.push(100);
Stack<Integer> copy = copyStack(stack);
System.out.println("Original Stack: " + stack);
System.out.println("Copy Stack: " + copy);
```

### 6. collapse

```python
public static void collapse(Stack<Integer> stack) {
    Queue<Integer> queue = new LinkedList<>();
    
    // Move from stack to queue
    while (!stack.isEmpty()) {
        queue.add(stack.pop());
    }
    
    // Move from queue back to stack for correct order  
    while (!queue.isEmpty()) {
        stack.push(queue.remove());
    }
    
    // Sort the stack by adding pairs of elements then add to queue
    while (!stack.isEmpty()) {
        int first = stack.pop();
        if (!stack.isEmpty()) {             // if stack not empty get the next element and find sum
            int second = stack.pop();
            queue.add(first + second);      // else just add the single element to the queue
        } else {
            queue.add(first);
        }
    }

    // Move from queue back to original stack for correct order
    while (!queue.isEmpty()) {
        stack.push(queue.remove());
    }
}
```

```python
Stack<Integer> stack = new Stack<>();
stack.push(1);
stack.push(2);
stack.push(3);
stack.push(4);
stack.push(5);

collapse(stack);
System.out.println("Collapsed Stack: " + stack);
```

### 7. reverseHalf

```python
public static void reverseHalf(Queue<Integer> queue) {
    Stack<Integer> stack = new Stack<>();
    int size = queue.size();
    
    // Odd position to stack, rotate even position to queue.back 
    for (int i = 0; i < size; i++) {
        int num = queue.remove();
        if (i % 2 == 0) { // even index
            queue.add(num);
        } else { // odd index
            stack.push(num);
        }
    }

    // Weave odd and even index back together
    while(!stack.isEmpty()) {
        queue.add(queue.remove()); // rotate the even index to the back of the queue
        queue.add(stack.pop()); // add the odd index from the stack to the back of the queue
    }

    // Handle odd size 
    if (size % 2 != 0) {
        queue.add(queue.remove()); // rotate the last even index to the back of the queue
    }
}
    
```

```python
Queue<Integer> queue = new LinkedList<>();
queue.add(1);
queue.add(8); 
queue.add(7);
queue.add(2); 
queue.add(9);
queue.add(18);
queue.add(12);
queue.add(0);

System.out.println("Original Queue: " + queue);
reverseHalf(queue);
System.out.println("Reversed Half Queue: " + queue);
```

### 8. reorder

```python
public static void reorder(Queue<Integer> queue){
    Stack<Integer> stack = new Stack<>();
    int size = queue.size();


    // Move negative number to stack
    for (int i = 0; i < size; i++) {
        int num = queue.remove();
        if (num < 0) {
          stack.push(num);
         } else {
          queue.add(num);
         }
    }   

    int positiveSize = queue.size();
    // Move negative from stack back to queue
    while (!stack.isEmpty()) {
        queue.add(stack.pop());
    }

    // Rotate negative number to front of queue
    for (int i = 0; i < positiveSize; i++) {
        queue.add(queue.remove());  
    }
        
}
```

```python
Queue<Integer> queue = new LinkedList<>();
queue.add(1);
queue.add(2); 
queue.add(-2);
queue.add(4); 
queue.add(-5);
queue.add(8);
queue.add(-8);
queue.add(12);
queue.add(-15);
queue.add(23);

System.out.println("Original Queue: " + queue);
reorder(queue);
System.out.println("Reordered Queue: " + queue);
```

### 9. compressDuplicates

```python
public static void compressDuplicates(Stack<Integer> stack) {
    Queue<Integer> queue = new LinkedList<>();
    
    // Flip order of the stack 
    while (!stack.isEmpty()) {
        queue.add(stack.pop());
    }
    while(!queue.isEmpty()) {
        stack.push(queue.remove());                    
    }

    // Check numbers and store the counts into queue
    int number = stack.pop();
    int count = 1;

    while (!stack.isEmpty()) {
        int next = stack.pop();
        if (next == number) {
            count++;
        } else {
            //reset the number and count 
            queue.add(count);
            queue.add(number);
            number = next;
            count = 1;
        }
    }
    // Add count and number
    queue.add(count);
    queue.add(number);


    // Restore the stack with the counts and numbers
    while (!queue.isEmpty()) {
        stack.push(queue.remove()); 
    }
}
```
