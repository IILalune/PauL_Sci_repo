### 2. Runtime #2

```python
int sum = 0;
for (int i = 1; i <= n * 2; i++) {
    for (int j = 1; j <= n; j++) {
        sum++;
    }
}
for (int j = 1; j < 100; j++) {
    sum++;
    sum++;
}
```

The run time for the code is O(N^2). 
The first for loop is nested loop so the runtime is N^2 and the next for loop has the run time of N. Together the runtime is N^2 + N of which with larger data set, the runtime would be 0(N^2)

### 6. Comparator

CaseInsensitiveComparator

```python
public class CasesensitiveComparator implements Comparator<String> {
    @Override
    public int compare(String s1, String s2) {
        return s1.compareToIgnoreCase(s2);
    }
}
```

DescendingComparator

```python
public class DescendingComparator implements Comparator<String> {
    @Override
    public int compare(String s1, String s2) {
        return -(s1.compareTo(s2));
    }
}
```

LengthComparator

```python
public class LengthComparator implements Comparator<String> {
    @Override
    public int compare(String s1, String s2) {
        return s1.length() - s2.length();
    }
}
```
