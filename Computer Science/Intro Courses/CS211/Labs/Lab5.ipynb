#### countLeftNodes

```python
public int countLeftNodes() {
    return count(overallRoot);
}

private int count(IntTreeNode root) {
    if (root == null) {
        return 0;
    } 
    int count = 0;
    
    // Count left node only
    if (root.left != null) {
        count = 1;
    }   
        // Traversal through tree, post-order
        return count + count(root.left) + count(root.right);
}
```

printLevel

```python
public void printLevel(int n) {
    // illegal cases
    if (n <= 0) {
        throw new IllegalArgumentException();
    }
    __printLevel(n, overallRoot);  
}
private void __printLevel(int height, IntTreeNode root) {
    if (root == null) {
        return;
    }
        // ||--> Stop at height 
    if (height == 1) {
        System.out.println(root.data);
    
    } else {
        // Recursive travesal through tree -->||
        __printLevel(height - 1, root.left);
        __printLevel(height - 1, root.right);
    }
}
```

printLeaves

```python
public void printLeaves() {
    if (overallRoot != null) {
        System.out.print("leaves: ");
        __printLeaves(overallRoot);
    } else {
        System.out.println("no leaves");
    }
}
private void __printLeaves(IntTreeNode root) {
    if (root == null) {
        return;
    }    
    if (root.right == null && root.left == null ) {
        System.out.print(root.data + " ");
    } else {
        __printLeaves(root.right);
        __printLeaves(root.left);
    }       
}
```
