### 1. largestNumber

```python
/**
 * Helper method for recursive case of getting the largest number in an integer array
 * @param array 	is the input array to be exam by the method to extract the largest 
 * 					integer presented in the array.
 * @param base		is the base case of the recursive method.
 * @return 	an integer that is the largest integer presented in the array structure. 
 */
private static int largestNumber_r(int[] array,int base) {
    // Reach base case, return the max
    if (base == array.length - 1) {
        return array[base];
    }
    // Recursive through array
    int max = largestNumber_r(array, base + 1);
        // Compare the max with the current element
        if (array[base] > max) {
            return array[base];
        } else {
            return max;
        }
}

/**
 * Method for recursive technique of getting the largest number in an integer array
 * @param array 	is the input array to be exam by the method to extract the largest 
 * 					integer presented in the array.
 * @return 	an integer that is the largest integer presented in the array structure. 
 */

public static int largestNumber(int[] array) {
    if (array.length == 0) {
        throw new IllegalArgumentException("Empty array!");
    }
    return largestNumber_r(array, 0);
    }

```

Test Case

```python
int[] array = {};
int[] array1 = {6, 3, 8};  
int[] array2 = {6, 3, 8, 1, 9};
int[] array3 = {6, 3, 8, 1, 9, -2};
int[] array4 = {-5, -16, 8, -3, -1, -2, 99};

System.out.println(largestNumber(array));
System.out.println(largestNumber(array1)); 
System.out.println(largestNumber(array2)); 
System.out.println(largestNumber(array3)); 
System.out.println(largestNumber(array4)); 
```

### 3 reverseLastK


```python

/**
 * Helper method for recursive case of getting the largest number in an integer array.
 * @param array 	The input array from the method to be reverse.
 * @param start     The starting index of the array to be reversed at k.
 * @param end		The end of the array list.
 * @return void.    The method will reverse the k+ elements of the array from the 
 *                  original array to the end of the array list.
 */

// Helper method for recursive with base case from k to end of list
public static void reverseLastK_r(int[] array, int start, int end) {
    
    // base case at the end of the array
    if (start >= end) {
        return;
    }

    // Recursive for reversing
    int temp = array[start];
    array[start] = array[end];
    array[end] = temp;

    reverseLastK_r(array, start + 1, end - 1);
}

/**
 * Method for recursive of reversing an array from a desired location in the array 
 * to the end of the same array list.
 * @param array 	The input array from the method to be reverse.
 * @param k
 * @return void      The array will reverse part of the desired location of the array 
 *                   to the end of the array list.
 */
// Recursive case, reverse from k to end. 
public static void reverseLastK(int[] array, int k) {
    // Boundary case for short arrays
    if (array == null || k <= 1 || array.length < 1) {
        return;
    }

    // Fix k if it is larger than array length
    if (k > array.length) {
        k = array.length;
    }

    // Recursvive       // count back ward from end for start at k
    reverseLastK_r(array, array.length - k, array.length - 1);      // (array, start, end)    
}    
```

Test case

```python
int[] array = {};
int[] array1 = {5};
int[] array2 = {1, 2, 3, 4, 5};
int[] array3 = {1, 2, 3, 4, 5, 6, 7, 8};

reverseLastK(array, 0);
reverseLastK(array1, 3);
reverseLastK(array2, 8);
reverseLastK(array3, 5);
System.out.println(Arrays.toString(array));
System.out.println(Arrays.toString(array1));
System.out.println(Arrays.toString(array2));
System.out.println(Arrays.toString(array3));
```
